# JUSTIFICACION DE LA SELECCION DE CLASES
La seleccion de estas 15 clases (botella, taza, tenedor, cuchillo, cuchara, tazon, manzana, sandwich, naranja, brocoli, zanahoria, perro caliente, pizza, dona y pastel) se basa en cubrir los tres pilares fundamentales de un entorno culinario: utensilios de consumo, herramientas de corte y grupos alimenticios variados. Al limitar el modelo YOLO solo a estos objetos, se reduce drásticamente el ruido visual y los falsos positivos que harian que el sistema desperdiciara recursos procesando objetos irrelevantes como muebles o personas en un contexto donde solo importa la interaccion con la comida.

# PROBLEMATICA
En entornos de cocina profesional, cafeterias de autoservicio o sistemas de asistencia domestica, la gestion manual de inventarios y el monitoreo de procesos son lentos y propensos al error humano. No existe un control en tiempo real que permita identificar que ingredientes se estan utilizando o que utensilios estan presentes en una estacion de trabajo sin intervencion manual. Los modelos de vision computacional generales suelen fallar al intentar procesar demasiadas categorias simultaneamente, perdiendo precision y velocidad de respuesta en hardware limitado.

# FORMA EN QUE SOLUCIONA
Este sistema especializado soluciona el problema mediante la automatizacion de la vigilancia y el conteo de activos en tiempo real. Al implementar el filtrado de clases en YOLO, el software puede:

    - Agilizar el proceso de cobro en cajas de autoservicio al detectar platos preparados (pizza, sandwich) y bebidas (botella, taza).

    - Monitorear la higiene y preparacion al identificar la presencia de herramientas especificas (cuchillo, tenedor).

    - Gestionar el stock de alimentos frescos (frutas y vegetales) de forma visual.
    La solucion ofrece una respuesta de baja latencia que permite generar alertas o registros de datos precisos sin necesidad de que un operador humano supervise cada movimiento en la cocina.

```python
from ultralytics import YOLO
import cv2

# 1. Cargar modelo YOLOv8
model = YOLO("yolov8n.pt")

# 2. IDs de clases de cocina
cocina_ids = [
    39, # bottle
    41, # cup
    42, # fork
    43, # knife
    44, # spoon
    45, # bowl
    47, # apple
    48, # sandwich
    49, # orange
    50, # broccoli
    51, # carrot
    52, # hot dog
    53, # pizza
    54, # donut
    55  # cake
]

# 3. Abrir cámara
cap = cv2.VideoCapture(0)

if not cap.isOpened():
    print("No se pudo abrir la cámara")
    exit()

while True:

    ret, frame = cap.read()
    if not ret:
        break

    # 4. Predicción filtrando clases
    results = model(frame, classes=cocina_ids, conf=0.3)

    result = results[0]

    # 5. Dibujar detecciones
    for box in result.boxes:

        cls_id = int(box.cls[0])
        conf = float(box.conf[0])

        x1, y1, x2, y2 = map(int, box.xyxy[0])

        label = f"{model.names[cls_id]} {conf:.2f}"

        cv2.rectangle(frame, (x1,y1), (x2,y2), (0,255,0), 2)
        cv2.putText(frame, label, (x1,y1-10),
                    cv2.FONT_HERSHEY_SIMPLEX, 0.6, (0,255,0), 2)

    # 6. Mostrar video
    cv2.imshow("Deteccion Cocina YOLOv8", frame)

    # salir con ESC
    if cv2.waitKey(1) == 27:
        break

cap.release()
cv2.destroyAllWindows()
```
