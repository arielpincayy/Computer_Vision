```python
from ultralytics import YOLO
import cv2

# 1. Cargar modelo YOLOv8
model = YOLO("yolov8n.pt")

# 2. IDs de clases de cocina
cocina_ids = [
    39, # bottle
    41, # cup
    42, # fork
    43, # knife
    44, # spoon
    45, # bowl
    47, # apple
    48, # sandwich
    49, # orange
    50, # broccoli
    51, # carrot
    52, # hot dog
    53, # pizza
    54, # donut
    55  # cake
]

# 3. Abrir cámara
cap = cv2.VideoCapture(0)

if not cap.isOpened():
    print("No se pudo abrir la cámara")
    exit()

while True:

    ret, frame = cap.read()
    if not ret:
        break

    # 4. Predicción filtrando clases
    results = model(frame, classes=cocina_ids, conf=0.3)

    result = results[0]

    # 5. Dibujar detecciones
    for box in result.boxes:

        cls_id = int(box.cls[0])
        conf = float(box.conf[0])

        x1, y1, x2, y2 = map(int, box.xyxy[0])

        label = f"{model.names[cls_id]} {conf:.2f}"

        cv2.rectangle(frame, (x1,y1), (x2,y2), (0,255,0), 2)
        cv2.putText(frame, label, (x1,y1-10),
                    cv2.FONT_HERSHEY_SIMPLEX, 0.6, (0,255,0), 2)

    # 6. Mostrar video
    cv2.imshow("Deteccion Cocina YOLOv8", frame)

    # salir con ESC
    if cv2.waitKey(1) == 27:
        break

cap.release()
cv2.destroyAllWindows()
```
